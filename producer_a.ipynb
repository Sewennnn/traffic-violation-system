```python
import pandas as pd
from time import sleep
from json import dumps
from kafka3 import KafkaProducer
import random
import datetime as dt


hostip = "10.192.101.151"

def publish_message(producer_instance, topic_name, data):
    """
    This entire document is referenced from week 11 tutorial.
    Publishes a single message to the specified Kafka topic.

    """
    try:
        producer_instance.send(topic_name, value=data)
        print('Message published successfully. Data: ' + str(data))
    except Exception as ex:
        print('Exception in publishing message.')
        print(str(ex))
        
        
def connect_kafka_producer():
    """
    Publishes a single message to the specified Kafka topic.

    """
    _producer = None
    try:
        _producer = KafkaProducer(bootstrap_servers=[f'{hostip}:9092'],
                                  value_serializer=lambda x: dumps(x).encode('ascii'),
                                  api_version=(0, 10))
    except Exception as ex:
        print('Exception while connecting Kafka.')
        print(str(ex))
    finally:
        return _producer
    
    
if __name__ == '__main__':
   
    topic = 'camera_event_171'
    print('Publishing records..')
    producer = connect_kafka_producer()
    
    df = pd.read_csv('camera_event_A.csv')
    grouped = df.groupby('batch_id')

    # Iterate over batches
    for batch_id, batch_df in grouped:
        records = batch_df.to_dict(orient='records')
        for record in records:

            publish_message(producer, topic, record)
        print(f'Batch {batch_id} sent. Sleeping for 5 seconds...\n')
        sleep(5)
       
```
